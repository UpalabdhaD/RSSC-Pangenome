```python

def read_tsv_as_string(file_path):
    with open(file_path, mode='r', newline='') as tsvfile:
        
        reader = csv.reader(tsvfile, delimiter='\t')
        
        lines = list(reader)

        if not lines:
            return ""  # Empty file

        return lines[1][4]

read_tsv_as_string(f"{kaks_res_dir}/group_11375.kaks.tsv")
```

```python
xx = [{os.path.basename(genefam).split(".")[0]: read_tsv_as_string(f"{kaks_res_dir}/{genefam}")} for genefam in kaks2_output]
```

```python
valid_kaks = []

for genefam in kaks2_output:
    kaks = read_tsv_as_string(f'{kaks_res_dir}/{genefam}')

    # Check if it's a non-empty, non-'NA' string and a valid number
    if isinstance(kaks, str) and kaks.strip() != '' and kaks.strip().upper() != 'NA':
        try:
            famname = os.path.basename(genefam).split(".")[0]
            kaks_value = float(kaks.strip())
            valid_kaks.append({"gene": famname, "kaks": kaks_value})
        except ValueError:
            # Skip if conversion to float fails
            continue
```

```python
len(valid_kaks)
```

```python
df = pd.DataFrame.from_dict(valid_kaks)
# df.

# df[df[['kaks']] != '' and df[['kaks']] != 'NaN']
import numpy as np

# Make sure NaNs are handled properly
df['kaks'] = pd.to_numeric(df['kaks'], errors='coerce')  # Converts invalid strings to NaN

# Filter out rows where 'kaks' is missing (NaN)
clean_df = df[~df['kaks'].isna()]
```

```python
clean_df.to_csv("<PROJECT_ROOT>/Results/DNDS/RSSC_kaks2_df.csv",
         index=False)
```
